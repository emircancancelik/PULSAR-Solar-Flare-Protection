```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
import joblib

# 1. Veriyi yükle (Filtrelenmiş elit seti kullanıyoruz)
df = pd.read_csv("ALAZ_Ultra_Elit_Egitim_Seti.csv")

# 2. Sadece 3 ana özelliği seç (App.py ile tam uyumlu)
features = ['kp_value', 'Yorunge_Egilimi_Inclo', 'Ortalama_Hareket_NoKozai']
target = 'BSTAR_MACRO'

X = df[features]
y = df[target]

print(f"🚀 Model Eğitiliyor... Giriş Sayısı: {X.shape[1]} (Beklenen: 3)")

# 3. Eğit ve Kaydet
model = RandomForestRegressor(n_estimators=50, max_depth=10, n_jobs=-1, random_state=42)
model.fit(X, y)

joblib.dump(model, "solar_flare_model.pkl")
print("✅ BAŞARILI: 'solar_flare_model.pkl' artık sadece 3 veri bekliyor.")
```
